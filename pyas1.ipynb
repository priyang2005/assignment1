```python
temp = float(input("Enter temperature (°C): "))

if temp < 15:
    print("Cold Alert")
    print("Action: Increase heating.")

elif temp <= 25:
    print("Normal")
    print("Action: No action required.")

elif temp <= 40:
    print("Warning")
    print("Action: Monitor temperature.")

else:
    print("Critical Alert")
    print("Action: Stop operations immediately.")
```

```python
weight = float(input("Enter weight (kg): "))
value = float(input("Enter order value: "))

if weight < 1:
    if value >= 50:
        cost = 0
    else:
        cost = 5

elif weight <= 5:
    if value >= 100:
        cost = 8
    else:
        cost = 12

elif weight <= 10:
    cost = 15

else:
    cost = 25 + (weight - 10) * 2

print("Shipping Cost = $", cost)
```

```python
bmi = float(input("Enter BMI: "))
bp = int(input("Enter Blood Pressure: "))
chol = int(input("Enter Cholesterol: "))

levels = ["Underweight", "Healthy", "Overweight", "Obese", "Severely Obese"]

if bmi < 18.5:
    risk = 0
elif bmi < 25:
    risk = 1
elif bmi < 30:
    risk = 2
elif bmi < 35:
    risk = 3
else:
    risk = 4

if bp > 140 or chol > 200:
    if risk < 4:
        risk += 1

print("Final Health Risk:", levels[risk])
```

```python
vehicle = input("Enter vehicle type (motorcycle/car/suv): ")
hours = int(input("Enter parking hours: "))
passholder = input("Daily pass holder? (yes/no): ")

if passholder == "yes":
    fee = 15

else:
    if vehicle == "motorcycle":
        fee = 1 + (hours - 1) * 0.5

    elif vehicle == "car":
        fee = 2 + (hours - 1) * 1

    elif vehicle == "suv":
        fee = 3 + (hours - 1) * 1.5

    if hours > 6:
        fee = fee + fee * 0.20

print("Parking Fee = $", fee)
```
